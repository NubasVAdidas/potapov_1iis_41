**ЗАДАНИЕ 4**

```python
age_mean = titanic.Age.mean()
print(f"Средний возраст пассажиров: {age_mean:.2f}")
print(f"\n\n{60*'*='}\n\n")
fare_min = titanic.Fare.min()
fare_max = titanic.Fare.max()
print(f"Максимальная цена билета: {fare_max:.2f}\nМинимальная цена билета: {fare_min}")
print(f"\n\n{60*'*='}\n\n")
age_std = titanic.Age.std()
print(f"Среднее отклонение возраста {age_std: .2f}")
```

**ЗАДАНИЕ 5**

```python
age_sort = titanic[titanic.Age > 30]
display(age_sort)
print(f"\n\n{60*'*='}\n\n")
fare_sort = titanic[titanic.Fare > titanic.Fare.mean()]
print(f"Средняя цена билета: {titanic.Fare.mean():.2f}")
display(fare_sort)
print(f"\n\n{60*'*='}\n\n")
rich_woman = titanic[(titanic.Pclass == 1) & (titanic.Sex == "female")]
display(rich_woman)
age_rw = rich_woman.Age.mean()
print(f"Средний возраст женщин в первом классе: {age_rw:.2f}")
```

**ЗАДАНИЕ 6**

```python
survived_people = titanic[(titanic.Survived == 1) & (titanic.Age < 18)]
display(survived_people)
sp_fare_mean = survived_people.Fare.mean()
sp_count = survived_people.PassengerId.count()
print(f"Средняя стоимость билета выживших мл. 18 = {sp_fare_mean:.2f}\nКол-во таких пассажиров: {sp_count:.2f}")
```

**ЗАДАНИЕ 7**

```python
class_g = titanic.groupby('Pclass')
apfagg_g = class_g.agg(
    Mean_Age = ("Age", "mean"),
    Passenger_Count = ("PassengerId", "count"),
    Fare_Mean = ("Fare", "mean")
)
sort_g = apfagg_g.sort_values(by="Fare_Mean",ascending=True)
display(sort_g)
```

**ЗАДАНИЕ 8**

**ЗАДАНИЕ 9**

```python

null_counts = titanic.isnull().sum()
null_percentages = (titanic.isnull().sum() / len(titanic) * 100).round(2)
sum_table = pd.DataFrame({
    'Null_Count': null_counts,
    'Null_%': null_percentages
})
display(sum_table)
print(f"\n\n{60*'*='}\n\n")
display(titanic.Embarked.value_counts())
titanic['Embarked'] = titanic['Embarked'].fillna('S')
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].mean())
print(f"\n\n{60*'*='}\n\n")
passg_dup = titanic.duplicated(subset=['PassengerId']).sum()
print(f"Количество дубликатов в PassengerId: {passg_dup}")

#Я хз зачем это делать если дубликатов нет, но раз в задании сказано удалить то я это делаю
titanic = titanic.drop_duplicates(subset=['PassengerId'], keep='first')

```

**ЗАДАНИЕ 10**

```python
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch'] + 1
titanic['IsAlone'] = titanic['FamilySize'].apply(lambda x: 1 if x == 1 else 0)
titanic['AgeGroup'] = pd.cut(titanic['Age'],
 bins=[0, 12, 18, 35, 60, 100],
labels=['ребёнок', 'подросток', 'молодой', 'взрослый',
'пожилой'])
display(titanic[['FamilySize', 'IsAlone', 'Age', 'AgeGroup']])
print(f"\n\n{60*'*='}\n\n")
age_group = titanic.pivot_table(
    values=['Survived', 'Fare', 'PassengerId'],
    index='AgeGroup',
    aggfunc={"Survived":"mean", "Fare":"mean", "PassengerId":"count"}
)
display(age_group)
```

**ЗАДАНИЕ 11**

```python
display(titanic.Embarked.value_counts())
print(f"\n\n{60*'*='}\n\n")
emb_group = titanic.groupby("Embarked").agg(
    Age_Mean = ("Age", "mean"),
    Fare_Mean = ("Fare", "mean"),
    Survived_Mean = ("Survived", "mean"),
    First_Class = ("Pclass", lambda x: (x == 1).mean())
)
display(emb_group)
```
