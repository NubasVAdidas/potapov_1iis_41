Шаг 2. Формированние корпуса

```python
corpus = """
Вчера я приехал в Пятигорск, нанял квартиру на краю города, на самом высоком месте, у подошвы Машука. Во время грозы облака будут спускаться до моей кровли.
Нынче в пять часов утра, когда я открыл окно, моя комната наполнилась запахом цветов, растущих в скромном палисаднике.
Ветки цветущих черешен смотрят мне в окна, и ветер иногда усыпает мой письменный стол их белыми лепестками.
Вид с трех сторон у меня чудесный. На запад пятиглавый Бешту синеет, как последняя туча умирающей бури.
На север подымается Машук, как мохнатая персидская шапка, и закрывает всю эту часть небосклона.
На восток смотреть мне веселее: внизу передо мною пестреет чистенький, новенький городок, шумят целебные ключи, шумит разноязычная толпа.
А там, дальше, амфитеатром громоздятся горы все синее и туманнее, а на краю горизонта тянется серебряная цепь снеговых вершин, начинаясь Казбеком и кончаясь двуглавым Эльборусом.
Весело жить в такой земле! Какое-то отрадное чувство разлито во всех моих жилах. Воздух чист и свеж, как поцелуй ребенка, солнце светит ярко, небо синее - чего бы, кажется, больше? Зачем тут страсти, желания, сожаления?
"""

print("Размер исходного корпуса:", len(corpus), "символов.")
```

Переменная corpus хранит сырые данные. Функция len() нужна просто для того, чтобы убедиться, что текст не пустой и имеет достаточный объем для обучения. Библиотека markovify автоматически разобьет этот текст на предложения по точкам и подготовит токены.

Шаг 3. Модель первого порядка

```python
model_order_1 = markovify.Text(corpus, state_size=1)
for i in range(3):
    sentence = model_order_1.make_sentence(tries=100)
    print(f"Фраза {i+1}: {sentence}")
```

Параметр state_size=1 задает жесткое ограничение памяти. Модель видит слово "он" и выбирает следующее слово только на основе того, что чаще всего шло после "он" в оригинале. Из-за этого грамматика часто ломается (согласования падежей теряются), но текст получается оригинальным.
Параметр tries=100 означает: если алгоритм зашел в тупик (нет пути к слову END) или сгенерировал фразу, которая на 100% совпадает с оригиналом (библиотека защищает от банального копирования), он начинает генерацию заново, и так до 100 раз.

Шаг 4. Модель второго порядка

```python
model_order_2 = markovify.Text(corpus, state_size=2)
for i in range(3):
    sentence = model_order_2.make_sentence(tries=100)
    print(f"Фраза {i+1}: {sentence}")
```

Здесь модель учитывает контекст из двух слов. Например, она видит "Печорин был" и понимает, что дальше с высокой вероятностью пойдет прилагательное в мужском роде. Грамматика становится правильной, но из-за узкого контекстного окна модель начинает просто выдергивать готовые куски предложений из оригинального текста Лермонтова.

Функция markovify.merge (или combine в зависимости от версии, в документации указано combine) берет матрицы вероятностей обеих моделей и смешивает их. Веса [0.5, 0.5] означают, что при выборе следующего слова алгоритм будет с равной вероятностью смотреть в граф новостей или в граф сказок. На выходе получается сюрреалистичный текст
